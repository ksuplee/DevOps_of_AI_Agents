# [실습 09] 사회적 규약(Commitment) 상태기계 구현

> **연계**: 제4부 09장(규범적 에이전트 팀) · **환경**: Google Colab · Python

**학습 목표**
- 에이전트 간 **약속(규약)** 의 생애주기(생성→활성→이행/위반)를 상태기계로 구현한다(09-2).
- 약속 위반을 감지해 책임을 추적하는 원리를 체험한다.

```python
class Commitment:
    def __init__(self, debtor, creditor, condition, result):
        self.debtor, self.creditor = debtor, creditor
        self.condition, self.result = condition, result
        self.state = "생성됨"
    def activate(self):
        if self.state == "생성됨":
            self.state = "활성"
    def fulfill(self):
        if self.state == "활성":
            self.state = "이행됨"
    def violate(self):
        if self.state == "활성":
            self.state = "위반됨"
    def __repr__(self):
        return f"C({self.debtor}→{self.creditor}: {self.result}) [{self.state}]"
```

## 1. 개발자→QA 약속: '테스트 가능한 빌드 제공'

```python
c = Commitment("개발자", "QA", "코드 완성", "테스트 가능한 빌드 제공")
print(c)
c.activate();  print(c)      # 조건 충족 → 활성
c.fulfill();   print(c)      # 결과 이행 → 완수
```

## 2. 위반 시나리오와 책임 추적

```python
team = [
    Commitment("아키텍트", "개발자", "설계 요청", "API 명세 전달"),
    Commitment("개발자", "QA", "빌드 요청", "테스트 빌드 제공"),
]
for c in team:
    c.activate()
team[1].violate()   # 개발자가 약속 위반

print("현재 상태:")
for c in team:
    print(" ", c)
violators = [c.debtor for c in team if c.state == "위반됨"]
print("책임(위반) 주체:", violators)
```
