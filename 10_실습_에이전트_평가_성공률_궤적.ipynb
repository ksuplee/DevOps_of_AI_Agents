# [실습 10] 에이전트 평가 — 성공률과 궤적 채점

> **연계**: 제5부 10장(평가) · **환경**: Google Colab · Python

**학습 목표**
- **End-to-End(성공률)** 와 **Trajectory(궤적)** 평가를 직접 계산한다(10-2).
- '결과는 맞았지만 과정이 비효율적'인 경우를 잡아내는 다차원 평가를 체험한다.

## 1. 에이전트 실행 로그(궤적) 예시

```python
runs = [
    {"goal": "날씨", "steps": ["search", "answer"], "success": True},
    {"goal": "환율", "steps": ["search", "search", "search", "answer"], "success": True},
    {"goal": "번역", "steps": ["answer"], "success": False},
]
OPTIMAL_STEPS = 2  # 이상적 단계 수
```

## 2. 다차원 지표 계산 (10-3)

```python
n = len(runs)
success_rate = sum(r["success"] for r in runs) / n
avg_steps = sum(len(r["steps"]) for r in runs) / n
# 궤적 효율: 이상 단계 대비 얼마나 낭비했나
efficiency = sum(min(1.0, OPTIMAL_STEPS/len(r["steps"])) for r in runs) / n

print(f"성공률(End-to-End) : {success_rate:.0%}")
print(f"평균 단계 수        : {avg_steps:.1f}")
print(f"궤적 효율(Trajectory): {efficiency:.0%}")

print("\n[진단]")
for r in runs:
    tag = "성공" if r["success"] else "실패"
    waste = "⚠️ 비효율" if len(r["steps"]) > OPTIMAL_STEPS else ""
    print(f"  {r['goal']}: {tag}, {len(r['steps'])}단계 {waste}")
```

## 3. 정리
- **성공률(결과)** 만 보면 놓치는 **궤적 비효율**을 함께 측정했다(10-1, 10-2).
- 좋은 평가는 결과·과정·효율을 함께 보는 **다차원**이다(10-3).
