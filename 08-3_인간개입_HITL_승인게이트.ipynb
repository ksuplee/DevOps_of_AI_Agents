# [실습 08] 인간 개입(HITL) — 위험 행동 승인 게이트

> **연계**: 제4부 08장(인간-에이전트 협업) · **환경**: Google Colab · Python

**학습 목표**
- **고위험 외부 행동 전 사람 승인**을 받는 게이트를 구현한다(08-2).
- 되돌릴 수 없는 행동(결제·삭제)만 사람에게 확인받는 원리를 체험한다.

## 1. 행동을 위험도로 분류

```python
HIGH_RISK = {"send_payment", "delete_file", "send_email"}

def needs_approval(action):
    return action in HIGH_RISK
```

## 2. 승인 게이트가 있는 실행기

저위험 행동은 자동 실행, 고위험 행동은 **사람의 승인**을 기다립니다. (Colab에서 input으로 시뮬레이션)

```python
def execute(action, arg, auto_approve=None):
    if needs_approval(action):
        print(f"⚠️ 고위험 행동 요청: {action}({arg})")
        if auto_approve is None:
            ans = input("승인하시겠습니까? (y/n): ")
        else:
            ans = "y" if auto_approve else "n"
            print("승인 여부(자동):", ans)
        if ans.lower() != "y":
            return f"❌ 거부됨: {action}"
    return f"✅ 실행됨: {action}({arg})"

print(execute("search", "날씨"))                 # 저위험 → 자동
print(execute("send_payment", "50000원", auto_approve=True))   # 고위험 → 승인 필요
print(execute("delete_file", "report.pdf", auto_approve=False))
```

## 3. 정리
- **되돌릴 수 없는 고위험 행동**만 사람 승인을 거치게 했다(08-2 개입 타이밍).
- 자율성 슬라이더: 저위험은 자율, 고위험은 승인형.
- **더 해보기**: `auto_approve=None`으로 바꿔 직접 y/n을 입력해 보세요.
