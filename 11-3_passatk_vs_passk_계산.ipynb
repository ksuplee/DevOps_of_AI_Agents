# [실습 11] pass@k vs pass^k — 잠재력과 신뢰성

> **연계**: 제5부 11장(벤치마크) · **환경**: Google Colab · Python

**학습 목표**
- 같은 실행 결과를 **pass@k(한 번이라도 성공)** 와 **pass^k(모두 성공)** 로 다르게 해석한다(11-2).
- '잘할 수 있다'와 '늘 잘한다'가 다름을 수치로 확인한다.

## 1. 반복 실행 결과 시뮬레이션

에이전트는 비결정적이라 같은 과제도 매번 결과가 다릅니다. 시도당 성공확률 p로 k번 실행합니다.

```python
def pass_at_k(p, k):
    # k번 중 적어도 1번 성공할 확률
    return 1 - (1 - p) ** k

def pass_hat_k(p, k):
    # k번 모두 성공할 확률
    return p ** k

p = 0.9  # 시도당 성공률 90%
print(f"시도당 성공률 p = {p}\n")
print(f"{'k':>3} | {'pass@k(잠재력)':>14} | {'pass^k(신뢰성)':>14}")
for k in [1, 2, 4, 8]:
    print(f"{k:>3} | {pass_at_k(p,k):>13.1%} | {pass_hat_k(p,k):>13.1%}")
```

## 2. 해석

- **pass@k** 는 k가 커질수록 **올라간다**(관대) → 잠재력.
- **pass^k** 는 k가 커질수록 **내려간다**(엄격) → 신뢰성.
- 90% 성공률이라도 8번 연속 모두 성공은 43%뿐! 실서비스엔 pass^k가 중요하다.

```python
# 실제 실행을 랜덤으로 시뮬레이션해 확인
import random
random.seed(0)
def simulate(p, k, trials=10000):
    at_k = hat_k = 0
    for _ in range(trials):
        outcomes = [random.random() < p for _ in range(k)]
        if any(outcomes): at_k += 1
        if all(outcomes): hat_k += 1
    return at_k/trials, hat_k/trials

a, h = simulate(0.9, 8)
print(f"시뮬레이션(k=8): pass@8≈{a:.1%}, pass^8≈{h:.1%}")
```

## 3. 정리
- 같은 결과도 **pass@k(낙관)** 와 **pass^k(엄격)** 로 정반대로 읽힌다(11-2).
- 프로덕션 판단에는 **신뢰성 지표(pass^k)** 를 반드시 본다.
