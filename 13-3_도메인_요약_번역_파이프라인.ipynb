```python
import torch
from transformers import pipeline
gen = pipeline("text-generation", model="Qwen/Qwen2.5-1.5B-Instruct",
               torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
               device_map="auto")
def task(system, user):
    msg = [{"role": "system", "content": system}, {"role": "user", "content": user}]
    return gen(msg, max_new_tokens=200, do_sample=False)[0]["generated_text"][-1]["content"].strip()
```

## 1. 문서 하나로 3가지 업무 자동 수행

```python
doc = '''인공지능 에이전트 시장은 2024년 이후 빠르게 성장하고 있다.
소프트웨어 개발, 신약 개발, 고객 서비스 등 다양한 분야에서 도입이 확산되고 있으며,
특히 코드 자동화와 문서 처리에서 생산성 향상이 두드러진다.'''

print("■ 요약\n", task("한국어 문서를 2문장으로 요약하라.", doc), "\n")
print("■ 핵심 키워드\n", task("핵심 키워드 5개를 콤마로만 출력하라.", doc), "\n")
print("■ 영문 번역(요약본)\n", task("다음을 영어로 번역하라.", task("한 문장 요약.", doc)))
```
